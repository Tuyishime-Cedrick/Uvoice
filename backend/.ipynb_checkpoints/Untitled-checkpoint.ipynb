```python
import matplotlib.pyplot as plt
import numpy as np
labels = "vote resultat:"
Pour = [40]
Contre = [-50]
neutre = [10]

plt.barh(labels, Pour, color = 'green')
plt.barh(labels, Contre, color = 'red' )
plt.barh(labels, neutre, color = 'blue')

plt.axvline('0')


plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
labels = "vote resultat:"
vote = np.array([50,10,40])
meslabels = ['Pour', 'Neutre', 'Contre']
myexplode = [0, 0.1, 0]
mycolors = ["green", "blue", "red"]
plt.pie(vote, labels = meslabels, explode = myexplode, shadow = True, colors = mycolors, autopct='%1.1f%%')
plt.legend(title = "resultats de vote:")

plt.show()
```
